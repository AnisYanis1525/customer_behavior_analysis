```python
# create column : age_group 

labels = ['young', 'adult', 'middle_age', 'old']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
frequency_mapping = {
    'Fortnightly' : 14,
    "Weekly" : 7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90
    }

df['frequency_purchases_days'] = df['frequency_of_purchases'].map(frequency_mapping)

```
